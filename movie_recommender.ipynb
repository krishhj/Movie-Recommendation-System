```python
import pandas as pd
import numpy as np
```

```python
movies = pd.read_csv('tmdb_5000_movies.csv')
credits = pd.read_csv('tmdb_5000_credits.csv')
```

```python
movies = movies.merge(credits, on='title')
```

```python
movies = movies[['genres', 'id', 'keywords','original_language', 'original_title', 'overview','popularity','cast', 'crew']]
```

```python
movies.dropna(inplace=True) #dropping null values, inplace means changes will be applied to original dataframe and inplace=True means it will not return anything
```

```python
movies.duplicated().sum() #checking for duplicate values
```

```python
#the genres column contains a list of dictionaries. We need to convert it into a list of strings. For that we will use ast library
import ast #ast is used to convert string representation of a list or dictionary into an actual list or dictionary
def convert(obj):
    L = []
    for i in ast.literal_eval(obj): #literal_eval is used to evaluate a string containing a Python literal or container display
        L.append(i['name']) 
    return L

def convertcharacter(obj):
    L = []
    for i in ast.literal_eval(obj)[:5]: 
        L.append(i['character']) 
        L.append(i['name']) 
    return L

def fetch_director_producer(obj):
    names = []
    crew_list = ast.literal_eval(obj)
    
    for person in crew_list:
        if person['job'] == 'Director':
            names.append(person['name'])
        elif person['job'] == 'Producer':
            names.append(person['name'])
        
    return list(set(names))
```

```python
movies['genres'] = movies['genres'].apply(convert) #applying the convert function to the genres column, apply function is used to apply a function to each element of a series
movies['keywords'] = movies['keywords'].apply(convert)
movies['cast'] = movies['cast'].apply(convertcharacter)
movies['crew'] = movies['crew'].apply(fetch_director_producer)
```

```python
movies['crew'][0]
```

```python
movies['overview'] = movies['overview'].apply(lambda x:x.split())
```

```python
movies['genres'] = movies['genres'].apply(lambda x:[i.replace(" ","") for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
movies['cast'] = movies['cast'].apply(lambda x:[i.replace(" ","") for i in x])
movies['crew'] = movies['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies['cast'] = movies['cast'].apply(lambda x: [name for item in x for name in item.split('/')])
```

```python
movies.head()
```

```python
movies['tags'] = movies['genres'] + movies['keywords'] + movies['overview'] + movies['cast'] + movies['crew']
```

```python
movies['tags'][0]
```

```python
new_df = movies[['id','original_title','original_language','tags']]
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x:" ".join(x))
```

```python
new_df.head()
```

```python
new_df['tags'][0]
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x: x.lower())
```

```python
new_df.head()
```

```python
new_df.info()
```

```python
#Now all the movies data is processed and We will now work on Vectorization
#Vectorization is the process of turning text into numbers.
# Think about it like this: computers and machine learning models don't understand words like "Action," "Adventure," or "Johnny Depp." They only understand numbers. Vectorization is the crucial step that translates human language into a numerical format that a computer can work with.
#In this case we will turn tags into vectors and based on the vectors we will recommend movies 

from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(stop_words='english',max_features=5000)
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()

def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    
    return " ".join(y)
```

```python
new_df['tags']= new_df['tags'].apply(stem)
```

```python
vectors = cv.fit_transform(new_df['tags']).toarray()
```

```python
np.set_printoptions(threshold=np.inf)
```

```python
cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
similarity[0]
```

```python
new_df[new_df['original_title'] == "Batman Begins"].index[0]
```

```python
sorted(enumerate(similarity[0]),reverse=True,key=lambda x:x[1])[1:11]
```

```python
def recommend(movie):
    try:
        movie_index = new_df[new_df['original_title'] == movie].index[0]
        distances = similarity[movie_index]
        movies_list = sorted(enumerate(distances), reverse=True, key=lambda x: x[1])[1:7]

        for i in movies_list:
            print(new_df.iloc[i[0]].original_title)
    except KeyError:
            print(f"Error: Movie '{movie}' not found in the dataset. Please check for exact spelling.")
    except Exception as e:
            print(f"An error occurred: {e}")
```
